## **Question 1: Define Data Quality in the context of ETL pipelines. Why is it more than just data cleaning?**

**Answer:**

**Data Quality** in ETL pipelines refers to the **accuracy, completeness, consistency, validity, and reliability** of data as it moves from source systems to the target data warehouse.

### Key Points:

* It ensures data is **fit for reporting and decision-making**
* Data quality checks are applied at **Extract, Transform, and Load** stages
* It focuses on **prevention of errors**, not just correction

### Why it is more than data cleaning:

* Data cleaning only fixes errors (nulls, duplicates)
* Data quality includes:

  * Validation against business rules
  * Consistency across systems
  * Timeliness and freshness of data
* Poor-quality data can still be “clean” but **incorrect**

 **Conclusion:**
Data quality is a **broader concept** that ensures trustworthy data, while data cleaning is only one part of it.

## **Question 2: Explain why poor data quality leads to misleading dashboards and incorrect decisions.**

### **Answer:**

Poor data quality directly affects the **accuracy of analytics and dashboards**, leading to wrong business decisions.

### **Reasons:**

* Incorrect values result in **wrong KPIs**
* Missing data shows **incomplete trends**
* Duplicate records inflate totals and revenue
* Inconsistent formats cause **calculation errors**

### **Impact on decisions:**

* Management may trust incorrect reports
* Financial planning becomes unreliable
* Business strategies are based on false insights

 **Example:**
If transaction amounts are missing or duplicated, total sales shown in a dashboard will be incorrect.

 **Conclusion:**
Poor data quality creates **false confidence**, which is more dangerous than no data at all.

## **Question 3: What is duplicate data? Explain three causes in ETL pipelines.**

**Answer:**

**Duplicate data** refers to records that appear more than once in a dataset, either exactly or partially.

### **Three common causes in ETL pipelines:**

1. **Multiple Data Sources**

   * Same record coming from different systems
   * No proper deduplication logic

2. **Reprocessing of Data**

   * ETL job reruns without clearing old data
   * Leads to repeated inserts

3. **Missing Unique Key**

   * No primary key or transaction ID
   * ETL cannot identify existing records

 **Conclusion:**
Duplicates distort reports and must be handled during transformation.

## **Question 4: Differentiate between exact, partial, and fuzzy duplicates.**

**Answer:**

| Type                  | Description                    | Example                                  |
| --------------------- | ------------------------------ | ---------------------------------------- |
| **Exact Duplicate**   | All fields are identical       | Same Txn_ID, Name, Amount                |
| **Partial Duplicate** | Some fields match, some differ | Same customer, different case or spacing |
| **Fuzzy Duplicate**   | Similar but not exact          | “Rahul Mehta” vs “Rahul M.”              |

### **Key Difference:**

* Exact → easy to detect
* Partial → needs rule-based matching
* Fuzzy → needs advanced matching logic

 **Conclusion:**
Understanding duplicate types helps choose the right deduplication strategy.

## **Question 5: Why should data validation be performed during transformation rather than after loading?**

**Answer:**

Data validation should be done during **transformation** to stop bad data before it enters the target system.

### **Reasons:**

* Errors are detected **early**
* Reduces rework and rollback
* Prevents corruption of target tables
* Improves ETL performance

### **Benefits:**

* Invalid data can be rejected or corrected
* Clean data reaches reporting layer
* Saves time and storage

 **Conclusion:**
Validating data during transformation ensures **high-quality, reliable data** in the final system.

## **Question 7: Write an SQL query on Sales_Transactions to list all duplicate keys and their counts using the business key (Customer_ID + Product_ID + Txn_Date + Txn_Amount).**

**Answer (SQL Query):**

```sql
SELECT
    Customer_ID,
    Product_ID,
    Txn_Date,
    Txn_Amount,
    COUNT(*) AS duplicate_count
FROM Sales_Transactions
GROUP BY
    Customer_ID,
    Product_ID,
    Txn_Date,
    Txn_Amount
HAVING COUNT(*) > 1;
```

**Explanation:**

* Groups data using the given business key
* `COUNT(*) > 1` finds duplicate records
* Shows duplicate keys along with their counts

## **Question 8: Enforcing Referential Integrity Assume the following Customers_Master tables:**
## **Create `Customers_Master` Table**

```sql
CREATE TABLE Customers_Master (
    CustomerID VARCHAR(10) PRIMARY KEY,
    CustomerName VARCHAR(50),
    City VARCHAR(30)
);
```
##  **Insert Data into `Customers_Master`**

```sql
INSERT INTO Customers_Master (CustomerID, CustomerName, City) VALUES
('C101', 'Rahul Mehta', 'Mumbai'),
('C102', 'Anjali Rao', 'Bengaluru'),
('C103', 'Suresh Iyer', 'Chennai'),
('C104', 'Neha Singh', 'Delhi');
```
```sql
SELECT * FROM Customers_Master;
```
**Task:**
Identify `Sales_Transactions.Customer_ID` values that violate referential integrity when joined with `Customers_Master` and write an SQL query to detect such violations.

### **Explanation:**

* Referential Integrity ensures that every `Customer_ID` in `Sales_Transactions` exists in `Customers_Master`.
* If a `Customer_ID` is not present in the master table, it creates a referential integrity violation.
* Such records can cause incorrect joins and inaccurate reports.
* These violations can be detected using a `LEFT JOIN`.

### **SQL Query to detect referential integrity violations:**

```sql
SELECT DISTINCT s.Customer_ID
FROM Sales_Transactions s
LEFT JOIN Customers_Master c
    ON s.Customer_ID = c.CustomerID
WHERE c.CustomerID IS NULL;
```
### **Result (from given data):**

* **C105**
* **C106**

 These `Customer_ID` values are not present in `Customers_Master`,
so they **violate referential integrity**.

### **Conclusion:**

> Referential integrity violations must be identified and handled to maintain accurate and reliable data.
